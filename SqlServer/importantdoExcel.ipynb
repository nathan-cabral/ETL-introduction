```python
# Importando o arquivo em excel
df = pd.read_excel(r"C:\Users\natha\Downloads\teste excell2.xlsx")

#df #-> mostra tudo
#df.head(10) #-> mostra a quantidade de linhas passada por parametro
df.head() #mostrar as 5 primeiras  linhas do arquivo
```

Manipular dados especficos

```python
#localizar um dado espeficico em uma coluna:
#df_calca = df.loc["coluna" ==  calca]
df_calca=df.loc[df["name"]=="calca"]
df_calca
```

```python
#localizar elemento com valor > que eu escolher
dfMaior100=df.loc[df["price"]>100]
dfMaior100
```

```python
#localizar elemento com valor < que eu escolher
dfMenor100=df.loc[df["price"]<=100]
dfMenor100
```

```python
#localizar elemento entre o valor que eu escolher
dfEntre100a300=df.loc[(df["price"]>=100) & (df["price"]<=300)]
dfEntre100a300
```

```python
#pegar itens diversos
itens=["calca","perfume","camisa"]
dfItens=df.loc[df["name"].isin(itens)]
dfItens
```
