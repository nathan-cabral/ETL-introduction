```python
import pandas as pd
```

Um arquivo CSV (Comma-Separated Values ou Valores Separados por Vírgula) é um arquivo de texto simples que armazena dados tabulares, onde cada linha representa um registro e os valores são separados por um delimitador, geralmente uma vírgula. 



```python
df=pd.read_csv(r"C:\Users\natha\Downloads\dados_etl.csv",delimiter=',')
df.head(30)
```

```python
# filtrar 1 item especifico de uma coluna
dfEstado=df.loc[df["state"]=="Bahia"]
dfEstado
```

```python
# Escolher estados específicos 
estados=["Rio de Janeiro","Santa Catarina","São Paulo"]
dfEstados=df.loc[df["state"].isin(estados)]
dfEstados
```
