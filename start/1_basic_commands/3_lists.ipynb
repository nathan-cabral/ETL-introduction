Principais metodos para listas

```python
listaExInt=[1,2,3,4,5,6,7,8,9]
listaExStrin=["natinho","chuva","cachorro","papi"]
```

```python
#append  -> Adiciona um elemento no final da lista
print(listaExInt)
listaExInt.append(10)
print(listaExInt)

print(listaExStrin)
listaExStrin.append("tartaruga")
print(listaExStrin)
```

```python
#clear  -> apaga todos os elementos da lista
print(listaExInt)
listaExInt.clear()
print(listaExInt)
```
